# 1-DARS: NumPy — Poydevor
## Nima uchun NumPy?
 * Python'ning oddiy list i sekin. NumPy esa C tilida yozilgan, shuning uchun million raqamni bir zumda hisoblaydi.

```python
import numpy as np
```

```python
# Oddiy Python list vs NumPy

oddiy = [1, 2, 3, 4, 5]
np_array = np.array([1, 2, 3, 4, 5])

# Har birini 2ga ko'paytir
# List: for loop kerak
# NumPy: bir qatorda!

oddiy_array = [son * 2 for son in oddiy]
np_array = np_array * 2

print("oddiy_array:",oddiy_array)
print("np_array:",np_array)
```

# Loyihaga kirish — Do'kon ma'lumotlari

```python
# Do'konimizda 7 kunlik sotuv (so'm)

sotuv = np.array(
    [
        1_200_000,  # Dushanba
        980_000,    # Seshanba
        1_450_000,  # Chorshanba
        870_000,    # Payshanba
        1_600_000,  # Juma
        2_100_000,  # Shanba
        1_350_000   # Yakshanba
    ]
)

kunlar = [
    "Dushanba",
    "Seshanba",
    "Chorshanba",
    "Payshanba",
    "Juma",
    "Shanba",
    "Yakshanba"
]

#  -------------- Asosiy hisob kitob -----------------

jami_sotuv = sotuv.sum()
ortacha_sotuv = sotuv.mean()
eng_yuqori_sotuv = sotuv.max()
eng_past_sotuv = sotuv.min()
std_devation = sotuv.std().round()  # <- qanchalik farq qiladi

print("Jami sotuv:", jami_sotuv)
print("O'rtacha sotuv:", ortacha_sotuv)
print("Eng yuqori sotuv:", eng_yuqori_sotuv)
print("Eng past sotuv:", eng_past_sotuv)
print("Std. devation:", std_devation)
```

# NumPy'ning kuchi — Filtering

```python
# Qaysi kunlar o'rtachadan yuqori sotilgan?
ortacha = sotuv.mean()
yaxshi_kunlar = sotuv[sotuv > ortacha]

print("Yaxshi kunlar soni:", len(yaxshi_kunlar))
print("Yaxshi kun sotuvlari:", yaxshi_kunlar)

# Indeks orqali - qaysi kun nomi?
yaxshi_kun_index = np.where(sotuv > ortacha)[0]
for i in yaxshi_kun_index:
    print(f" {kunlar[i]}: {sotuv[i]:,} so'm")
```

# Shape, Dimension — Ko'p o'lchamli massivlar

```python
# Bir haftadagi 3 ta mahsulot sotuvlari
# Har qator = bir kun, har ustun = bir mahsulot
mahsulotlar = np.array([
    [120, 45, 30],   # Dushanba
    [98,  60, 25],   # Seshanba
    [145, 30, 50],   # Chorshanba
    [87,  70, 20],   # Payshanba
    [160, 55, 60],   # Juma
    [210, 90, 80],   # Shanba
    [135, 40, 45],   # Yakshanba
])

print("Shape:", mahsulotlar.shape)  # (7, 3) - 7 kun, 3 mahsulot

# Har mahsulotning haftalik jami (ustun bo'yicha yig'indi)
print("Har mahsulotning jami:", mahsulotlar.sum(axis=0))   # (axis=0) ⬇️ ustunlar bo'yicha  (axis=1)➡️qatorlar bo'yicha qo'shadi

# Har kunning jami savdosi (qator bo'yicha)
print("Har kun jami:", mahsulotlar.sum(axis=1))
```

# Vazifa

```python
# 1. Bu massivni yaratish
temperaturalar = np.array([22, 25, 19, 28, 31, 27, 24])

# Quyidagilarni hisoblaymiz

# a) O'rtacha temperatura
ortacha_temperatura = temperaturalar.mean()

# b) Eng issiq kun (indeksi ham)
eng_issiq_kun = temperaturalar.max()
eng_issiq_indeks = temperaturalar.argmax()

# c) 25 dan yuqori kunlar soni
yuqori_kunlar = temperaturalar[temperaturalar > 25]

# d) Barcha temperaturalarni Fahrenheitga o'tkazish: F = C * 9/5 + 32
fahrenheit = temperaturalar * 9/5 + 32

print("O'rtacha temperatura:", ortacha_temperatura)
print(f"Eng issiq kun: {eng_issiq_kun}°C, {eng_issiq_indeks}-indeks")
print("25 dan yuqori kunlar soni:", len(yuqori_kunlar))
print("Temperaturalar Fahrenheitga o'tkazilganda:", fahrenheit)
```
