# 2-DARS: Pandas — Ma'lumotni jadvalga sol
## Endi NumPy massivlarimizni real jadvalga aylantiramiz.

```python
import numpy as np
import pandas as pd
```

```python
# O'tgan darsdan ma'lumotlar

kunlar = ["Du", "Se", "Cho", "Pa", "Ju", "Sha", "Ya"]
sotuv  = [1_200_000, 980_000, 1_450_000, 870_000, 1_600_000, 2_100_000, 1_350_000]
mijozlar = [45, 38, 52, 31, 60, 87, 49]
harorat  = [22, 25, 19, 28, 31, 27, 24]

# DataFrame yaratish - bu bizning "jadvalimiz"
df = pd.DataFrame({
    "Kun"      : kunlar,
    "Sotuv"    : sotuv,
    "Mijozlar" : mijozlar,
    "Harorat"  : harorat
})

print(df)
```

## Umumiy ma'lumot

```python
df.info()
df.describe()
```

## Ustunni tanlash

```python
df["Kun"]         # bir ustun → Series
```

```python
df[["Kun", "Sotuv"]]  # bir nechta ustun → DataFrame
```

# Qatorni tanlash

```python
df.iloc[0] # 0-indeks (Dushanba)
```

```python
df.iloc[-1] # oxirgi qator (Yakshanba)
```

```python
df.iloc[2:5] # 2 dan 4 gacha
```

# Yangi ustun qo'shish

```python
df["Mijoz_boshiga"] = df["Sotuv"] / df["Mijozlar"]

print(df)
```

```python
df["Mijoz_boshiga"] = df["Mijoz_boshiga"].round()

print(df)
```

```python
# Kun "yaxshimi yoki yo'q" - o'rtachadan yuqorimi
ortacha = df["Sotuv"].mean()
df["Yaxshi_kun"] = df["Sotuv"] > ortacha

df["Yaxshi_kun"] = df["Yaxshi_kun"].replace({
    True: "Yaxshi",
    False: "Yomon"
})

print(df)
```

# Groupby — Guruhlash

```python
# Yaxshi va yomon kunlarni solishtir
df.groupby("Yaxshi_kun")["Sotuv"].mean()
```

# Vazifa
  ## Quyidagilarni qilamiz:

  * 1. df dan faqat Shanba va Yakshanba qatorlarini chiqar (iloc yoki shart bilan)

  * 2. Yangi ustun qo'sh: "Sotuv_darajasi"
    - Sotuv > 1_500_000 bo'lsa → "Yuqori"
    - Sotuv > 1_000_000 bo'lsa → "O'rta"
    - Qolganlar          → "Past"
    - Maslahat: np.select() yoki df["ustun"].apply() ishlatib ko'r

  * 3. Har bir "Sotuv_darajasi" guruhida
    - nechta kun borligini hisобла
    - Maslahat: .value_counts()

  ## Bonus savol: iloc va loc farqi nima?

```python
# Faqat Shanba va Yakshanba kunlar
df.loc[5:6]
```

```python
# Yangi ustun: "Sotuv_darajasi"

natija = [
    df["Sotuv"] >= 1_500_000,
    df["Sotuv"] >= 1_000_000,
]

daraja = ["Yuqori", "O'rta"]


df["Sotuv_darajasi"] = np.select(
    natija,
    daraja,
    default="Past"
)

print(df)
```

```python
# Har bir "Sotuv_darajasi" guruhida nechta kun borligi

df["Yaxshi_kun"].value_counts()
```

# iloc va loc farqi?

* Pandasda ikkalasi ham ma’lumot tanlash uchun ishlatiladi.
  - loc  -> index/column nomi bilan => loc[0:2] (2 ham kiradi)
  - iloc -> index raqami bilan      => iloc[0:2] (2 kirmaydi)
